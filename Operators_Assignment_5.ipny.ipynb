30.Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.

```python
import math
def scientific_calculator():
    while True:
        print("1. Square Root")
        print("2. Factorial")
        print("3. Logarithm (base 10)")
        print("4. Power")
        print("5. Sine")
        print("6. Cosine")
        print("7. Tangent")
        print("8. GCD")
        print("9. LCM")
        print("10. Exit")

        example = int(input("Enter your choice: "))

        if example == 1:
            num = float(input("Enter number: "))
            print("Square Root =", math.sqrt(num))

        elif example == 2:
            num = int(input("Enter number: "))
            print("Factorial =", math.factorial(num))

        elif example == 3:
            num = float(input("Enter number: "))
            print("Logarithm =", math.log10(num))

        elif example == 4:
            base = float(input("Enter base: "))
            power = float(input("Enter power: "))
            print("Result =", math.pow(base, power))

        elif example == 5:
            degree = float(input("Enter angle in degrees: "))
            radian = math.radians(degree)
            print("Sine =", math.sin(radian))

        elif example == 6:
            degree = float(input("Enter angle in degrees: "))
            radian = math.radians(degree)
            print("Cosine =", math.cos(radian))

        elif example == 7:
            degree = float(input("Enter angle in degrees: "))
            radian = math.radians(degree)
            print("Tangent =", math.tan(radian))

        elif example == 8:
            num1 = int(input("Enter first number"))
            num2 = int(input("Enter second number"))
            print("GCD =", math.gcd(num1, num2))

        elif example == 9:
            num1 = int(input("Enter first number"))
            num2 = int(input("Enter second number"))
            lcm = abs(num1 * num2) // math.gcd(num1, num2)
            print("LCM =", lcm)

        elif example == 10:
            print("Calculator Closed")
            break
        else:
            print("Invalid Choice!")

scientific_calculator()
```

Expected Output

```python
# Function to calculate the factorial of a number
def calculate_factorial(number):
    if number < 0:
        return None              # Check for negative input
    factorial = 1 # Initialize factorial to 1
    for i in range(1, number + 1):# Calculate factorial using a loop
        factorial = factorial * i
    return factorial    # Return the factorial

# Main function
def main():
    number = int(input("Enter a number: ")) # Accept input from the user
    result = calculate_factorial(number)  # Call the function
    if result is None:  # Display the result
        print("Error! Factorial is not defined for negative numbers.")
    else:
        print("Factorial:", result)


# Execute the program
if __name__ == "__main__":
    main()
```

Return its reversed version.
Solve it first using slicing.
Write another function that reverses it using a loop.
Example Input
Enter text: Python
Expected Output
Reversed text: nohtyP

```python
def reverse_string(text):
  return text[::-1] # reverse using slicing
def reverse_string_loop(text): # creating function to reverse using loop
    reversed_text = ""  # Initialize an empty string
    # Traverse the string from the last character to the first
    for i in range(len(text) - 1, -1, -1):
        reversed_text += text[i]

    # Return the reversed string
    return reversed_text
def main():
    text = input("Enter text: ")  # # Accept input from the user
    reversed_by_slicing = reverse_string(text)  # Reverse using slicing
    reversed_by_loop = reverse_string_loop(text)  # Reverse using a loop
    print("Reversed text using slicing", reversed_by_slicing)
    print("Reversed text using loop", reversed_by_loop)
if __name__ == "__main__":
    main()
```

```python
def analyze_text(text): # define function with atgument text
    vowels = "aeiouAEIOU" #assiging both captial and small
    result = {
        "Total characters": len(text),
        "Total words": len(text.split()),
        "Total vowels": 0,
        "Total digits": 0,
        "Total spaces": 0
    }
    for ch in text:
        if ch in vowels:
            result["Total vowels"] += 1
        if ch.isdigit():
            result["Total digits"] += 1
        if ch == " ":
            result["Total spaces"] += 1
    return result
# Example usage
text = "Python 3 is Awesome 2026"
analysis = analyze_text(text)
for key, value in analysis.items():
    print(f"{key}: {value}")
```

```python
def get_even_numbers(start, end):
    if start >= end:
        return "Error= Starting number must be smaller than the ending number."

    even_numbers = []
    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers                                      # Example usage
print(get_even_numbers(1, 10))
print(get_even_numbers(8, 20))
print(get_even_numbers(10, 5))
```

```python
def create_bank_account(balance):
    def deposit(amount):
        nonlocal balance
        balance += amount
        print("Deposited:", amount)

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance!")

    def check_balance():
        print("Current Balance:", balance)

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()

deposit(500)
check_balance()

withdraw(300)
check_balance()

withdraw(1500)
check_balance()
```

```python

def display_employees():  # 1Display Employees
    for emp in employees:
        print(emp)

def display_names():  #2 Employee Names
    names = list(map(lambda emp: emp["name"], employees))
    print(names)

def high_performers():  # 3 High Performers
    result = list(filter(lambda emp: emp["performance_score"] >= 80, employees))
    for emp in result:
        print(emp)

def search_department():  #4 Search by Department
    dept = input("Enter Department: ")
    result = list(filter(lambda emp: emp["department"].lower() == dept.lower(), employees))
    if result:
        for emp in result:
            print(emp)
    else:
        print("Department not found.")

def revised_salaries(): #5 Revised Salaries
    revised = list(map(lambda emp: {
        "name": emp["name"],
        "salary": emp["salary"],
        "revised_salary":
            emp["salary"] * 1.15 if emp["performance_score"] >= 90 else
            emp["salary"] * 1.10 if emp["performance_score"] >= 80 else
            emp["salary"] * 1.05 if emp["performance_score"] >= 70 else
            emp["salary"]
    }, employees))

    for emp in revised:
        print(emp)

def salary_expense(): #6 Salary Expense
    revised = list(map(lambda emp: {
        "salary": emp["salary"],
        "revised_salary":
            emp["salary"] * 1.15 if emp["performance_score"] >= 90 else
            emp["salary"] * 1.10 if emp["performance_score"] >= 80 else
            emp["salary"] * 1.05 if emp["performance_score"] >= 70 else
            emp["salary"]
    }, employees))
    current = reduce(lambda total, emp: total + emp["salary"], employees, 0)
    revised_total = reduce(lambda total, emp: total + emp["revised_salary"], revised, 0)

    print("Current Salary :", current)
    print("Revised Salary :", revised_total)
    print("Additional Expense :", revised_total - current)

def highest_paid(): # 7.Highest Paid Employee
    emp = reduce(lambda e1, e2: e1 if e1["salary"] > e2["salary"] else e2,employees)
    print(emp)

def sort_employees(): #8. Sort Employees

    print("1. Name")
    print("2. Salary")
    print("3. Performance Score")
    print("4. Experience")

    choice = int(input("Enter choice: "))

    if choice == 1:
        result = sorted(employees, key=lambda emp: emp["name"])
    elif choice == 2:
        result = sorted(employees, key=lambda emp: emp["salary"])
    elif choice == 3:
        result = sorted(employees, key=lambda emp: emp["performance_score"])
    elif choice == 4:
        result = sorted(employees, key=lambda emp: emp["experience"])
    else:
        print("Invalid Choice")
        return

    for emp in result:
        print(emp)

def department_report(): # 9.Department Report
    dept = input("Enter Department: ")
    dept_emp = list(filter(lambda emp: emp["department"].lower() == dept.lower(), employees))
    if len(dept_emp) == 0:
        print("Department not found.")
        return
    count = len(dept_emp)
    total = sum(emp["salary"] for emp in dept_emp)
    avg_salary = total / count
    avg_score = sum(emp["performance_score"] for emp in dept_emp) / count
    report = {
        "Department": dept,
        "Employees": count,
        "Total Salary": total,
        "Average Salary": avg_salary,
        "Average Performance": avg_score
    }
    print(report)
while True:
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")

    option = int(input("Enter your choice"))
    if option == 1:
        display_employees()
    elif option == 2:
        display_names()
    elif option == 3:
        high_performers()
    elif option == 4:
        search_department()
    elif option == 5:
        revised_salaries()
    elif option == 6:
        salary_expense()
    elif option == 7:
        highest_paid()
    elif option == 8:
        sort_employees()
    elif option == 9:
        department_report()
    elif option == 10:
        print("thank you")
        break
    else:
        print("invalid")
```

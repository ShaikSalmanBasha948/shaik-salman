```python
def count_data_types(*values): # defining function using values
    integer_count = 0 # initializing values
    string_count = 0
    list_count = 0
    other_count = 0
    for value in values:
        if type(value) is int: # Check if the value is exactly an integer
            integer_count += 1
        elif isinstance(value, str):
            string_count += 1
        elif isinstance(value, list):
            list_count += 1
        else:
            other_count += 1   #If none of the above conditions match, increment other_count.
    print("Integers:", integer_count) # to print the all initialized values using print statement
    print("Strings:", string_count)
    print("Lists:", list_count)
    print("Others:", other_count)

count_data_types(10, "Python", [1, 2], 20, "SQL", 3.5, True) # assigning values
```

```python
def separate_even_odd(*numbers):  # define function using numbers
    even_numbers = []  # creating empty lists
    odd_numbers = []

    for number in numbers: #looping each numer in numbers
        if number % 2 == 0: #Check whether the number is even using number % 2 == 0
            even_numbers.append(number) #if the number is even, add it to the even_numbers list using append().
        else:
            odd_numbers.append(number) #otherwise, add it to the odd_numbers list using append().

    return even_numbers, odd_numbers  #After processing all numbers, return both lists.

evens, odds = separate_even_odd(10, 15, 22, 31, 40, 47) # call the functions

print("Even numbers:", evens)
print("Odd numbers:", odds)
```

```python
def build_select_query(table, *columns, **filters):#Define the function with a normal argument (table), *columns to accept any number of column names, and **filters to accept filter conditions
    selected_columns = ", ".join(columns) if columns else "*" #Check whether any column names are provided

    query = f"SELECT {selected_columns} FROM {table}" #Create the SQL query

    if filters:         #check whether any filter conditions are provided in
        conditions = []

        for key, value in filters.items():    #if filters exist, create an empty list named
            conditions.append(f"{key}='{value}'")#Loop through each key-value pair

        query += " WHERE " + " AND ".join(conditions)

    return query

query = build_select_query(
    "Employee",
    "name",
    "salary",
    department="IT",
    city="Bangalore"
)

print(query)
```

```python
def log_function_call(func):  #Define a decorator function that accepts another function (func) as an argument.
    def wrapper(*args, **kwargs):#Inside the decorator, define a wrapper function using *args and **kwargs to accept any positional and keyword arguments.
        print("Function name:", func.__name__)#Print the original function's name
        print("Positional:", args) #Print all positional arguments received by the wrapper
        print("Keyword:", kwargs) #Print all keyword arguments received by the wrapper
        result = func(*args, **kwargs) #Call the original function

        print("Result:", result) #Store the returned value in a variable named result

        return result

    return wrapper #Return the result from the wrapper function.

@log_function_call  #Apply the decorator to another function
def calculate_bill(price, quantity, discount=0):
    total = price * quantity
    discount_amount = total * discount / 100
    return total - discount_amount


calculate_bill(500, 3, discount=10)
```

```python
def classify_numbers(*numbers, **kwargs): #Define the function using *numbers to accept multiple numbers and **kwargs to accept keyword arguments
    lower_limit = kwargs.get("lower_limit", 0) # read the lowwr limit val,ue using get
    upper_limit = kwargs.get("upper_limit", 100)# read the upoper limit value

    below =[]  # creating empty lists
    within =[]
    above =[]

    for number in numbers: # using for loop compare the all numbers are mention in functions
        if number < lower_limit:
            below.append(number)
        elif number > upper_limit:
            above.append(number)
        else:
            within.append(number)

    return {    # assigned names using return keyword
        "below": below,
        "within": within,
        "above": above
    }

classification=classify_numbers(  # assigning the values bt calling the function
    5,
    15,
    25,
    35,
    45,
    lower_limit=20,
    upper_limit=40
)

print(classification)
```

```python
def merge_dictionaries(*dictionaries,**overrides): #Define the function using *dictionaries to accept multiple dictionaries and **overrides to accept keyword arguments
    merged={} # create a empty dictionary
                #using forloop
    for dictionary in dictionaries: #Check whether the current item is a dictionary
        if isinstance(dictionary,dict): #If it is a dictionary, merge it into merged using the update() method.
            merged.update(dictionary)

    merged.update(overrides)  #If a key already exists, the new value replaces the old value

    return merged #Return the final merged dictionary
result = merge_dictionaries(  #all the function with multiple dictionaries and keyword arguments
    {"name": "Rahul", "city": "Bangalore"},
    {"department": "IT", "experience": 3},
    {"experience": 4},
    city="Hyderabad",
    salary=80000
)

print(result)
```

```python
import math #Import the math module to use math.ceil() for rounding upward
def etl_summary(source, destination, *tables, **kwargs): #define the function
    row_counts = kwargs.get("row_counts", {}) # read the row count using kwarge
    batch_size = kwargs.get("batch_size", 10000)

    table_count = len(tables) # count the number of tables

    total_rows = sum( # caluclate total rowa by looping
        row_counts.get(table, 0)
        for table in tables
    )
    batches = math.ceil(total_rows / batch_size) if total_rows else 0 #caluclate requeried batches

    print("Source:", source)
    print("Destination:", destination)      #Print source, destination, table names, table count, total rows, and required batches
    print("Tables:", tables)
    print("Table count:", table_count)
    print("Total rows:", total_rows)
    print("Required batches:", batches)

etl_summary(      #Call the function with source, destination, table names, row counts, and batch size
    "SQL Server",
    "ADLS Gen2",
    "Customer",
    "Orders",
    "Products",
    row_counts={
        "Customer": 12000,
        "Orders": 55000,
        "Products": 3000
    },
    batch_size=10000
)
```

```python
def check_permissions(*actions, **kwargs):# define a function using actions and kwargs
    role_permissions = {  # create a dictionary by assigning some values
        "admin": {"read", "write", "update", "delete"},
        "editor": {"read", "write", "update"},
        "viewer": {"read"}
    }

    role = kwargs.get("role", "viewer") # read the user role
    allowed_actions = role_permissions.get(role, set()) # get the allowed actions for thast role

    print("User:", kwargs.get("username", "Unknown"))  # print the u8ser name
    print("Role:", role)

    for action in actions:
        if action in allowed_actions: # check weather the auction exisist
            print(action, ": ALLOWED")
        else:
            print(action, ": DENIED")

check_permissions( # assign values by calling the functions
    "read",
    "update",
    "delete",
    username="Rahul",
    role="editor"
)
```

```python
def process_order(order_id,*products,**kwargs): # define a function
    subtotal =0   # initialization
    total_quantity =0

    for product in products: #loop throiugh every product
        if not isinstance(product,dict):
            continue
        price = product.get("price",0)
        quantity = product.get("quantity",0)

        product_total = price * quantity

        subtotal += product_total
        total_quantity += quantity

    discount_percentage = kwargs.get("discount",0)
    tax_percentage = kwargs.get("tax",0)

    discount_amount = subtotal * discount_percentage / 100
    amount_after_discount = subtotal - discount_amount

    tax_amount = amount_after_discount * tax_percentage / 100
    final_amount = amount_after_discount + tax_amount

    print("Order ID:", order_id)
    print("Customer:", kwargs.get("customer", "Guest"))
    print("Payment mode:", kwargs.get("payment_mode", "Not provided"))
    print("Total quantity:", total_quantity)
    print("Subtotal:", round(subtotal, 2))
    print("Discount amount:", round(discount_amount, 2))
    print("Tax amount:", round(tax_amount, 2))
    print("Final amount:", round(final_amount, 2))

process_order(
    "ORD-1001",
    {
        "name": "Laptop",
        "price": 50000,
        "quantity": 1
    },
    {
        "name": "Mouse",
        "price": 1000,
        "quantity": 2
    },
    {
        "name": "Keyboard",
        "price": 2000,
        "quantity": 1
    },
    customer="Rahul",
    payment_mode="UPI",
    discount=10,
    tax=18
)
```
